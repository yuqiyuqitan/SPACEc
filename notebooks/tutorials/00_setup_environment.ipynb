## 1. Create folders

```python
from pathlib import Path

base = Path("spacec_work")  # change if wanted
data_path = base / "example_data" / "raw"
output_dir = base / "output"
for p in (data_path, output_dir):
    p.mkdir(parents=True, exist_ok=True)

print(f"data_path = '{data_path}/'")
print(f"output_dir = '{output_dir}/'")
```

## 2. Download example data

Download `example_data.zip` (414 MB) from [Dryad](https://doi.org/10.5061/dryad.brv15dvj1) and put it into `spacec_work/`. The cell below unzips it to `spacec_work/example_data/` (`raw/` and `processed/`).

```python
import zipfile

# unzip the download (zip already contains the example_data/ folder)
for z in base.glob("*.zip"):
    with zipfile.ZipFile(z) as f:
        f.extractall(base)

expected = [
    data_path / "channelnames.txt",
    data_path / "tonsil_tma.tif",
    base / "example_data" / "processed" / "adata_nn_demo_tonsil.h5ad",
]
missing = [str(p) for p in expected if not p.exists()]
print("OK, data found." if not missing else f"Missing: {missing}")
```
