## About This Notebook

**The SPACEc Style Configurator** is an interactive tool for customizing the visual appearance of spatial analysis plots in SPACEc. Whether you're preparing figures for publication, creating presentations, or simply want consistent styling across your analyses, this tool lets you adjust colors, fonts, sizes, and other visual elements in real-time.

### Key Features
- **Live Preview**: See changes instantly on four example plots as you adjust settings
- **Interactive Controls**: No coding required—use sliders, dropdowns, and color pickers
- **One-Click Export**: Save your custom style as a reusable configuration file
- **Template-Based**: Start from built-in templates and customize to your needs

---

## How to Use This Tool

### Step 1: Run the Configurator
Execute the cell below to launch the interactive style editor interface.

### Step 2: Customize Your Style
The interface displays adjustable parameters organized by category:
- **Colors**: Change plot color palettes, markers, backgrounds
- **Typography**: Adjust font sizes and styles for labels and titles
- **Layout**: Modify spacing, marker sizes, line widths
- **Other**: Additional visual settings

As you adjust each parameter using the controls, the four example plots update in real-time, showing you exactly how your changes affect the visualization.

### Step 3: Preview Your Changes
The example plots demonstrate:
- How your style looks on different plot types
- Color contrast and readability
- Overall visual coherence

Iterate on settings until you're satisfied with the appearance.

### Step 4: Save Your Style
Once you're happy with your customization:
1. Enter a file path in the **Output Path** field (e.g., `/path/to/my_style.mplstyle`)
2. Click **Save Style File**
3. The configuration is saved as a `.mplstyle` file

### Step 5: Use Your Style in Analysis
To apply your saved style to future analyses:
```python
import matplotlib.pyplot as plt
plt.style.use('/path/to/my_style.mplstyle')
```

## Common Workflows

**Publication-Ready Figures**
- Adjust colors for color-blind accessibility
- Increase font sizes for readability in print
- Reduce marker sizes for clarity when overlaying many points
- Save and reuse consistently across all figure panels

**Dark Mode Theme**
- Change background colors to dark
- Adjust text colors for contrast
- Reduce brightness of accent colors
- Perfect for presentations and dashboards

**High-Contrast Style**
- Maximize color differences between categories
- Increase line widths and marker sizes
- Use bold fonts for clarity
- Ideal for accessibility and visibility from a distance

---

## Tips for New Users

- **Start Simple**: Begin by changing one parameter at a time so you understand its effect
- **Test Colors**: Use the live preview to check that your color choices have good contrast
- **Preserve Templates**: If you like the default, save it with a different name before customizing
- **Experiment Freely**: You can always restart—no changes are permanent until you save
- **Use Consistent Settings**: Define a style once and apply it to all your figures for visual consistency
